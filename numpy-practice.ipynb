# Indexing, slicing and iterating

```python
a = np.arange(10)**3
a
```

```python
a[2]
```

```python
a[2:5]
```

```python
a[:6:2] = 1000
```

```python
a
```

```python
a[::-1]
```

```python
for i in a:
    print(i**(1 / 3.))
```

```python
def f(x, y):
    return 10 * x + y

b = np.fromfunction(f, (5, 4), dtype=np.int_)
b
```

```python
b[2, 3]
```

```python
b[0:5, 1] # each row in the second column of b
```

```python
b[:, 1] # equivalent to the previous example
```

```python
b[1:3, :]
```
